```python
import pandas as pd
import numpy as np

IN_CSV  = "../data stuff/augmented_telemetry_filtered_with_location.csv"
OUT_CSV = "../data stuff/next24h_consumption.csv"
TS_COL  = "recorded_at"
LOC_COL = "location_id"
DISP_COL= "dispensed_l"

df = pd.read_csv(IN_CSV, parse_dates=[TS_COL])
df[DISP_COL] = df[DISP_COL].fillna(0.0).astype(float)

df['date_only'] = df[TS_COL].dt.date
num_days = df['date_only'].nunique()

print(f"Total days found in dataset: {num_days}")

agg = df.groupby(LOC_COL, as_index=False).agg(
    total_week_l = (DISP_COL, 'sum')
)

agg['predicted_demand_l'] = agg['total_week_l'] / num_days

agg['recorded_at'] = pd.Timestamp(df[TS_COL].min())
agg = agg[[LOC_COL, 'recorded_at', 'predicted_demand_l']]

agg.to_csv(OUT_CSV, index=False)

print("\n--- New Ground Truth (Average 24h Demand) ---")
print(agg.head())
print(f"\nSaved to: {OUT_CSV}")
```
